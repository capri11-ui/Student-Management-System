```python
students = [{"ID": "093", "name": "Carissa P.", "balance": 1306.23},
            {"ID": "013", "name": "Sara H.", "balance": 894.52},
            {"ID": "298", "name": "Victor G.", "balance": 756.09}]
print(f"{'Welcome to the Roster!'}")
print(f"{'Here are the students:'}")
print(f"{'ID':<5}{'name':<10}{'balance':>10}")
for entry in students:
  print(f"{entry['ID']:<5}{entry['name']:<10}{entry['balance']:>10.2f}")
```

```python
def display_roster():
    print("Here are the students")
    print("ID Name Balance")
    for student in students:
        print(f"{student['ID']} {student['name']} {student['balance']:.2f}")

```

```python
def add_student():
    student_id = input("ID: ").strip()
    name = input("name: ").strip()
    while True:
        try:
            balance = float(input("balance: ").strip())
            break
        except ValueError:
            print("Invalid input. Enter a valid number for balance.")

    students.append({"ID": student_id, "name": name, "balance": balance})
    print("Student added!===========================")
```

```python
def pay_tuition():
    display_roster()
    while True:
        try:
            index = int(input("Index: ").strip())
            if index < 1 or index > len(students):
                print(f"Invalid index. Enter a number between 1 and {len(students)}.")
                continue
            amount = float(input("Amount: ").strip())
            students[index-1]["balance"] = max(0.0, students[index-1]["balance"] - amount)
            print("Student updated!===========================")
            break
        except ValueError:
            print("Invalid input. Enter numbers only.")
```

```python
def clear_zero_balance():
    global students
    students = [student for student in students if student["balance"] > 0]
    print("Students with $0 balance cleared!===========================")
```

```python
def main():
    while True:
        print("Welcome to Roster!")
        display_roster()
        print("(1) to add a student")
        print("(2) to pay tuition")
        print("(3) to clear")
        print("(4) to exit")
        choice = input("Your choice: ").strip()
        print()
        if choice == "1":
            add_student()
        elif choice == "2":
            pay_tuition()
        elif choice == "3":
            clear_zero_balance()
        elif choice == "4":
            print("See you later!")
            break
        else:
            print("Invalid choice. Try again.")
```

```python
if __name__ == "__main__":
  main()
```
